```python
df_tags_columns.columns = [f'tag{i+1}' for i in range(df_tags_columns.shape[1])]
```

## Concatenation of the tags columns with the original dataframe

```python
import geopy
```

```python
geolocator = geopy.Nominatim(user_agent="geoapi")
def get_address_info(address_name:str):
    location = geolocator.geocode(address_name)
    if location:
        elements = location.address.split(',')
        if len(elements) < 6:
            print(elements)
            pass
        return elements[-5].strip(), elements[-2].strip(), elements[-1].strip()
```

```python
privateless_df['city'], privateless_df['postalCode'],privateless_df['country'] = zip(*privateless_df['address'].apply(get_address_info))
```

```python
privateless_df
```

## Merging dataframe with coordinates values contained in another file

```python
geo_js = json.load(open('mapstr.geojson'))
```

```python
features = geo_js['features']
```

```python
df_rows = []
```

```python
for feature in features:
    properties = feature['properties']
    geometry = feature['geometry']

    name = properties.get('name', '')
    address = properties.get('address', '')
    icon = properties.get('icon', '')

    tags = [tag['name'] for tag in properties.get('tags', [])]

    coordinates = geometry.get('coordinates', [])

    df_row = {
        'name': name,
        'address': address,
        'icon': icon,
        'coordinates': coordinates
    }

    for i, tag in enumerate(tags):
        df_row[f'tag{i + 1}'] = tag

    df_rows.append(df_row)
```

```python
df_geo = pd.DataFrame(df_rows)
```

```python
df_geo
```

## Separation of longitude and latitude coordinates

```python
def format_coordinates(row:str):
    longitude = row[0]
    latitude = row[1]
    return longitude, latitude
```

```python
df_geo['longitude'], df_geo['latitude'] = zip(*df_geo['coordinates'].apply(format_coordinates))
```

```python
df_merged = pd.merge(df, df_geo, on='name', how='left')
```

```python
df_merged = df_merged.drop(columns=['address_y', 'icon_y', 'tag1_y', 'tag2_y', 'tag3_y', 'tag4_y', 'tag5_y', 'tag6_y', 'tag7_y'])
```

```python
df_merged.columns = [column[:-2] if column.endswith('_x') else column for column in df_merged.columns]
```

```python
df_merged
```

## Formatting of address column not well formatted by earlier function

```python
def address_formatting(address:str):
    parts = address.split()
    if len(parts) < 3:
        return
    if address.endswith('France'):
        code_postal = parts[-3]
        pays = parts[-1]
        ville = " ".join(parts[-2:-1])
        return code_postal, ville, pays
    if address.startswith('Centre'):
        code_postal = parts[-1]
        ville = " ".join(parts[-2:-1])
        return code_postal, ville, "France"
```

```python
df_merged[['Code Postal', 'Ville', 'Pays']] = df_merged['address'].apply(lambda x: pd.Series(address_formatting(x)))
df_merged['city'] = df_merged['city'].fillna(df_merged['Ville'])
df_merged['postalCode'] = df_merged['postalCode'].fillna(df_merged['Code Postal'])
df_merged['country'] = df_merged['country'].fillna(df_merged['Pays'])
df_merged = df_merged.drop(columns=['Code Postal', 'Ville', 'Pays'])
```
